```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Upload the file
uploaded = files.upload()

# Load the data
df = pd.read_csv("company_financials.csv")
print(df.head())
```

```python
# 1. Handle missing values (fill using average of neighboring periods)
df = df.interpolate(method='linear', limit_direction='both')

# 2. Add growth rate columns
df['Sales_Growth'] = df['Sales'].pct_change()
df['Net_Profit_Growth'] = df['Net_Profit'].pct_change()

# 3. Create Profit_Trend column vs previous period
# Using 0 as the threshold since >0 means growth
df['Profit_Trend'] = np.where(df['Net_Profit_Growth'] > 0, 'Profit Grew', 'Profit Declined')

# Drop the first row for any modeling/growth metrics since it will be NaN due to pct_change()
df_clean = df.dropna().copy()

# Save the cleaned dataset for your Streamlit app later
df_clean.to_csv("company_financials_clean.csv", index=False)
```

```python
# 1. Compute Statistics
stats_df = df_clean[['Sales', 'Net_Profit', 'OPM_Percent']].agg(['mean', 'median', 'std'])
print("Key Statistics:\n", stats_df)

# 2. Build Charts
# Chart A: Sales & Net_Profit trend line
plt.figure(figsize=(10, 5))
plt.plot(df_clean['Period'], df_clean['Sales'], marker='o', label='Sales')
plt.plot(df_clean['Period'], df_clean['Net_Profit'], marker='s', label='Net Profit')
plt.title('Sales & Net Profit Trend')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()

# Chart B: OPM_Percent bar chart
plt.figure(figsize=(10, 5))
sns.barplot(x='Period', y='OPM_Percent', data=df_clean, color='skyblue')
plt.title('OPM Percent per Period')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Chart C: Box plot of Net_Profit by Profit_Trend
plt.figure(figsize=(6, 5))
sns.boxplot(x='Profit_Trend', y='Net_Profit', data=df_clean)
plt.title('Net Profit Distribution by Profit Trend')
plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import warnings
warnings.filterwarnings('ignore')

# Define features and target
X = df_clean[['Sales_Growth', 'OPM_Percent', 'Interest', 'Other_Income']]
y = df_clean['Profit_Trend']

# Note: Due to the small sample size (approx 11 rows), we will train and test on the same data
# just to demonstrate the code functionality as requested by the assignment's constraints.
# Model 1: Logistic Regression
lr_model = LogisticRegression()
lr_model.fit(X, y)
lr_acc = accuracy_score(y, lr_model.predict(X))

# Model 2: Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X, y)
dt_acc = accuracy_score(y, dt_model.predict(X))

# Compare in a small table
model_comparison = pd.DataFrame({
    'Model': ['Logistic Regression', 'Decision Tree'],
    'Accuracy': [lr_acc, dt_acc],
    'Limitation Note': ['Trained on small sample size', 'Trained on small sample size']
})
print(model_comparison)
```
